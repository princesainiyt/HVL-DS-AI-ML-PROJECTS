### Join, Split, Sort and Filter

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

combined = np.concatenate((a, b))
combined
```

```python
combined = np.concatenate((b, a))
combined
```

```python
split_data = np.array_split(combined, 3)
split_data
```

```python
split_data = np.array_split(combined, 2)
split_data
```

```python
split_data = np.array_split(combined, 6)
split_data
```
